```python
#convert 0 & 1 in senior citizen to yes or no

def conv(value):
    if value == 1:
        return "yes"
    else:
        return "no"

df['SeniorCitizen'] = df['SeniorCitizen'].apply(conv)
```

```python
ax = sns.countplot(x = df['Churn'],data = df )

ax.bar_label(ax.containers[0])

plt.title("Count of Customers by Churn")
plt.show()
```

```python
# see percentage

plt.figure(figsize = (3,3))

churn_count = df['Churn'].value_counts()

plt.pie(
    churn_count,
    labels = churn_count.index,
    autopct = "%1.2f%%" 
    )

plt.title("Percentage of churned customers")
plt.show()
```

```python
plt.figure(figsize = (4,4))

sns.countplot(x = 'SeniorCitizen' , data = df ,hue = "Churn") 

plt.title('Churn by Senior Citizen')
plt.show()
```

```python
plt.figure(figsize=(2,2))

ct = pd.crosstab(df['SeniorCitizen'], df['Churn'])

# Convert each SeniorCitizen group into percentages
ct_percent = ct.div(ct.sum(axis=1), axis=0) * 100

ax = ct_percent.plot(
    kind='bar',
    stacked=True,
    figsize=(4,4)
)

plt.title('Churn by Senior Citizen (%)')
plt.xlabel('Senior Citizen')
plt.ylabel('Percentage')
plt.legend(title='Churn')

# Show percentage inside each section
for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.1f%%',
        label_type='center'
    )

plt.show()
```

```python
#create subplots 

cat_cols = [
    'gender',
    'SeniorCitizen',
    'Partner',
    'Dependents',
    'PhoneService',
    'MultipleLines',
    'InternetService',
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies',
    'Contract',
    'PaperlessBilling',
    'PaymentMethod',
    'Churn'
]
```

```python
fig, axes = plt.subplots(5, 4, figsize=(20, 25))

axes = axes.flatten()

for i, col in enumerate(cat_cols):
    sns.countplot(x=col, data=df, ax=axes[i])
    axes[i].set_title(f'{col}')
    axes[i].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

```

```python
fig, axes = plt.subplots(5, 4, figsize=(20, 25))

axes = axes.flatten()

for i, col in enumerate(cat_cols):
    sns.countplot(x=col, data=df, ax=axes[i] ,hue = df['Churn'])
    axes[i].set_title(f'{col}')
    axes[i].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

```

#Observation: Most customers who did not churn have PhoneService and InternetService. For OnlineSecurity, more customers who use it seem to stay. For OnlineBackup, TechSupport, and StreamingTV, customers who don't use these services show more churn. Overall, the plots show that service usage has some effect on customer churn.

```python
#Payment Method

plt.figure(figsize = (6,4))

ax = sns.countplot(x = "PaymentMethod",data = df ,hue ="Churn")
ax.bar_label(ax.containers[0])

plt.title('Count of Customers by payment method')
plt.xticks(rotation = 45)
plt.show()         
```

#Customer is likely to churn when he is using electonic check as a payment method


#Key Insights:
1. 26.54% of the customers have churned.
2. Gender does not seem to have much effect on churn.
3. Senior citizens have a higher churn proportion compared to non-senior citizens.
4. Customers with low tenure are more likely to leave, while long-term customers tend to stay.
5. Month-to-month customers have more churn than one-year and two-year contract customers.
6. Customers who do not use services such as OnlineSecurity, OnlineBackup and TechSupport show more churn.
7. Electronic check has comparatively higher churn than other payment methods.
8. Contract type, tenure, payment method and additional services seem to be the main factors related to churn.
